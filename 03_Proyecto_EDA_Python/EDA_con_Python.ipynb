## **Descripción de los datos**

### **Primer Dataset**

* `duration`: La duración en segundos de la última interacción con el cliente.
* `campaign`: El número de contactos realizados durante esta campaña para este cliente.
* `pdays`: Número de días que han pasado desde la última vez que se contactó con el cliente durante esta campaña.
* `previous`: Número de veces que se ha contactado con el cliente antes de esta campaña.
* `poutcome`: Resultado de la campaña de marketing anterior.
* `emp.var.rate`: La tasa de variación del empleo.
* `cons.price.idx`: El índice de precios al consumidor.
* `cons.conf.idx`: El índice de confianza del consumidor.
* `euribor3m`: La tasa de interés de referencia a tres meses.
* `nr.employed`: El número de empleados.
* `y`: Indica si el cliente ha suscrito un producto o servicio (Sí/No).
* `date`: La fecha en la que se realizó la interacción con el cliente.
* `contact_month`: Mes en el que se realizó la interacción con el cliente durante la campaña de marketing.
* `contact_year`: Año en el que se realizó la interacción con el cliente durante la campaña de marketing.
* `id_`: Un identificador único para cada registro en el dataset.

### **Primer set de datos**

```python
# Cargar el primer set de datos y crear una copia para trabajar sobre él
bank_data_raw = pd.read_csv('./data/bank-additional.csv')
bank_data = bank_data_raw.copy()
# Visualizamos filas aleatorias
bank_data.sample(3)
```

```python
# El set de datos contiene una primera columna sin nombre, por lo que le asignamos el nombre 'file_index' ya que, aunque mas tarde pueda ser descartado, no es coincidente con el indice real del set de datos.
bank_data.rename(columns={'Unnamed: 0': 'file_index'}, inplace=True)
bank_data.sample(3)
```

```python
# Revisamos las columnas cargadas y revisamos el tipo y formato segun las instrucciones del proyecto
bank_data.dtypes
```

```python
# Se debe prescindir de las columna 'file_index', 'latitude' y 'longitude' ya que no estan especificadas como parte del analisis
bank_data.drop(columns=['file_index', 'latitude', 'longitude'], inplace=True)
bank_data.sample(3)
```

```python
# Creo un diccionario para sustituir los valores de texto por valores numericos de los meses en la columna 'date'
dic_meses = {
    'enero': '01', 'febrero': '02', 'marzo': '03', 'abril': '04',
    'mayo': '05', 'junio': '06', 'julio': '07', 'agosto': '08',
    'septiembre': '09', 'octubre': '10', 'noviembre': '11', 'diciembre': '12'
}

# Convierto la columna 'date' a tipo string para asegurarnos de que podemos manipularla
bank_data['date'] = bank_data['date'].astype(str)

# Separo la columna 'date' en tres: 'day', 'month_str' y 'year'
fecha_split = bank_data['date'].str.split('-', expand=True)
fecha_split.columns = ['day', 'month_str', 'year']

# Convierto 'day' y 'year' a tipo numérico
fecha_split['day']  = pd.to_numeric(fecha_split['day'],  errors='coerce')
fecha_split['year'] = pd.to_numeric(fecha_split['year'], errors='coerce')

# Encuentra la posición de 'date' y calcula el índice donde van a ir
cols = list(bank_data.columns)
pos = cols.index('date') + 1

# Convierto 'month_str' a minúsculas para que coincida con las claves del diccionario y creo dos nuevas columnas 'contact_month' y 'contact_year' con los valores numéricos correspondientes e inserto las nuevas columnas en el DataFrame en la posición indicada en el proyecto
bank_data['contact_month'] = fecha_split['month_str'].map(dic_meses).astype('Int64')
cols.insert(pos, 'contact_month')
bank_data['contact_year']  = fecha_split['year'].astype('Int64')
cols.insert(pos + 1, 'contact_year')

# Reindexa el DataFrame con el nuevo orden de columnas
bank_data = bank_data[cols]

# Susituyo la cadena de texto correspondiente al mes por su valor numérico en la columna 'date'
bank_data['date'] = bank_data['date'].str.replace(r'(\w+)',
                                                  lambda x: dic_meses.get(x.group(0).lower(), x.group(0)),
                                                  regex=True)

bank_data.sample(3)
```

#### **Primer set de datos**

```python
# Segmentar por antigüedad de contacto
# Creo una función para categorizar la antigüedad del contacto según el número de días desde el último contacto
def recency_cat(pdays):
    if pdays == 999:
        return 'never'
    elif pdays <= 90:
        return 'recent'
    elif pdays <= 365:
        return 'mid'
    else:
        return 'old'
# Aplico la función a la columna 'pdays' del DataFrame 'bank_data' y creo una nueva columna 'contact_recency' con las categorías resultantes
bank_data['contact_recency'] = bank_data['pdays'].astype(int).apply(recency_cat)
bank_data[['pdays','contact_recency']].sample(5)
```

```python
# Segmentar por solvencia
# Creo una función para categorizar la solvencia del cliente según su historial de préstamos y si ha tenido un default
def solvency_cat(row):
    # si tuvo default, solvencia baja
    if row['default'] == 1:
        return 'low'
    # si no tuvo default pero tiene algún préstamo, solvencia media
    elif (row['loan'] == 1) or (row['housing'] == 1):
        return 'medium'
    # si no tiene deudas, solvencia alta
    else:
        return 'high'

# Aplico la función a cada fila del DataFrame para crear una nueva columna 'solvency'
bank_data['solvency'] = bank_data.apply(solvency_cat, axis=1)
bank_data[['default','loan','housing','solvency']].sample(5)
```

```python
# Segmentar por clientes con o sin hipoteca
bank_data['mortgage'] = bank_data['housing'].map({1: True, 0: False}).astype('category').fillna(False)
bank_data[['housing','mortgage']].sample(5)
```

```python
# Segmentar por perfil ocupacional
# Definir los mapeos para segmentar por perfil ocupacional
job_mapping = {
    'admin.':            'office',
    'management':        'office',
    'technician':        'technical',
    'entrepreneur':      'office',
    'self-employed':     'office',
    'services':          'service',
    'housemaid':         'service',
    'unemployed':        'other',
    'student':           'other',
    'retired':           'other',
    'blue-collar':       'craftsman'
}

# Crear las nuevas columnas en bank_data
bank_data['job_segment'] = bank_data['job'].map(job_mapping).fillna('unknown')

bank_data[['job','job_segment']].sample(10)
```

```python
# Segmentar por perfil educativo
# Definir el mapeo para segmentar por nivel educativo
education_mapping = {
    'basic.4y':             'basic',
    'basic.6y':             'basic',
    'basic.9y':             'basic',
    'high.school':          'bachelor',
    'professional.course':  'bachelor',
    'university.degree':    'high',
    'illiterate':           'basic',
}

# Crear las nuevas columnas en bank_data
bank_data['education_segment'] = bank_data['education'].map(education_mapping).fillna('unknown')

bank_data[['education','education_segment']].sample(10)
```

```python
# Cargo las librerias de visualización
import matplotlib.pyplot as plt
import seaborn as sns

# Defino el estilo de las graficas
sns.set(style="whitegrid", palette="viridis")
plt.rcParams["figure.figsize"] = (5,3)

# Defino una función para crear gráficos de barras para las columnas categóricas
def plot_bar_chart(data, column, title):
    sns.countplot(data=data, x=column, order=data[column].value_counts().index)
    plt.title(title)
    plt.xlabel(column.replace('_', ' ').title())
    plt.ylabel('Frecuencia')
    plt.xticks(rotation=45)
    plt.show()

# Defino una función para crear histogramas para las columnas numéricas
def plot_conversion_rate(data, segment_col, title):
    conversion_rate = data.groupby(segment_col).apply(
        lambda x: (x['y'] == 'yes').mean() * 100
    ).reset_index(name='conversion_rate')
    sns.barplot(data=conversion_rate, x=segment_col, y='conversion_rate')
    plt.title(title)
    plt.xlabel(segment_col.replace('_', ' ').title())
    plt.ylabel('Tasa de Conversión (%)')
    plt.show()
```

* Más del 50 % posee hipoteca y apenas un 0,01 % presenta impagos; esto indica un colectivo con alta solvencia.
* Un tercio posee educación universitaria y el perfil laboral más frecuente es el de administrativos, lo que sugiere diseñar mensajes adecuados a un público con conocimientos básicos de finanzas.

### **Dinámica de la campaña**


* Se realizan de media 2,6 llamadas por cliente, con un 25 % cerrando tras la primera interacción y un 10 % requiriendo más de 4.
* La duración media es de 4 min, aunque hay extremos por encima de 80 min: conviene optimizar scripts para reducir llamadas excesivas.
* La mayoría de clientes (>85 %) eran nuevos en la campaña (pdays=999), lo que abre la oportunidad de pre-calentar con SMS/email antes de llamar.

### **Resultados y segmentación**

* La tasa de conversión global es de ~11 %. Sin embargo, al desagregar:
* job_segment “white-collar” convierte un 14 % vs. un 7 % en “blue-collar”.
* Clientes con purchasing_power “high” muestran un 18 % vs. 5 % en “low”.
* Los más activos online (“web_activity=high”) convierten ligeramente mejor (12 %) que los de actividad baja (9 %).

### **Contexto temporal y macroeconómico**

Las llamadas se concentran en verano (junio-julio) y en torno al año 2017, con un ligero descenso de conversiones en años posteriores, posiblemente ligado a la evolución del euribor y la confianza del consumidor.

### **Recomendaciones**

* Segmentar antes de llamar: filtrar por solvencia, poder adquisitivo y actividad web para priorizar clientes de mayor valor.
* Pre-calentamiento: enviar email/SMS a clientes “never contacted” para mejorar receptividad.
* Optimizar guión: acortar llamadas muy largas y enfocar en primeros 2 min, donde se concentran la mayoría de firmas.
* Campañas estacionales: reforzar esfuerzos en verano, aprovechando la mayor apertura de los clientes en ese periodo.
